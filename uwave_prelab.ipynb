First, the ratio $\eta = E / H$ for a plane electromagnetic wave in vacuum can be obtained by solving Maxwell's equations, in particular Faraday's and Ampere's law, for the travelling plane wave solution. From Griffiths 9.2.1:
1. Fundamental plane wave solutions are expressed as complex exponentials with wave-number $k$ and angular frequency $\omega$. The right-hand rule triplet is $k, E, B$. Partial derivatives with respect to $x$ and $t$ spit out a multiple $ik$ and $i\omega$ respectively. (In 3-D space $k$ is generally a vector indicating direction of propagation, but I will use 1-D notation for simplicity.)
2. The propagation speed of the wave is $c=(\epsilon_0 \mu_0)^{-1/2}$ (Griffiths eq. 9.42).
3. Both Faraday and Maxwell imply a condition on the relative amplitude of the wavenumber. Ampere gives $k E_0 = \omega B_0$. So $E_0 / B_0 = \omega / k = \frac 1 c$ (Griffiths 9.46)
4. $H = B / \mu$ and in a vacuum $\mu = \mu_0$. So multiplying the amplitude condition by $\mu_0$ gets the final answer: $\eta = \mu_0 / c = \sqrt{\frac {\mu_0} {\epsilon_0}} \approx 376.73 \Omega$ 

Second, the reflection ratio $\Gamma$ between two media with different propagation impedances $\eta_1 \ne \eta_2$ is obtained by solving the boundary conditions on the $E$ and $H$ fields at the interface between the two media. Consider a 1-D setup, with an incident rightwards-travelling plane wave hitting the interface, partially reflecting, and partially reflecting, with the amplitudes of their electric fields being $A, B, C$. Assume the interface has no free charges which can create a surface current.
1. Obtain two equations from continuity of the E and H fields perpendicular to the interface (Griffiths eq. 7.62, 7.63).
2. Aligning the phases of the incident and reflected waves at the interface so that their electric fields add, the magnetic fields cancel because the propagation directions are opposite. (right-hand rule triplet). This allows the jump in impedance across the interface.

The two equations are:
$$A e^{-j(\omega t - ky)} + B e^{-j(\omega t + ky)} = C e^{-j(\omega t - ky)}$$

$$\frac{A}{\eta_1} e^{-j(\omega t - ky)} - \frac{B}{\eta_1} e^{-j(\omega t + ky)} = \frac{C}{\eta_2} e^{-j(\omega t - ky)}$$

Divide both equations by $A$ and let $\Gamma = \frac B A$ and $T = \frac C A$. Then after eliminating T, $\Gamma$ is found to be:

$$\Gamma = \frac{\frac{n_2}{n_1} - 1}{\frac{n_2}{n_1} + 1}$$

## Plot a Smith Chart

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
def f_gamma(z, z0=50):
    return (z - z0) / (z + z0)
```

```python
r_iter=[0, 50, 100, 300]
x_iter=[-200, -30, 0, 20, 50, 100, 300]
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect('equal')

# Googled: 
# 3. Create a custom formatter function for the imaginary axis
def imaginary_formatter(val, pos):
    if val == 0:
        return "0"
    return f"{val:g}j"  # Use "i" or "j" depending on your engineering/math preference

# Apply the formatter to the y-axis
ax.yaxis.set_major_formatter(ticker.FuncFormatter(imaginary_formatter))


for r in r_iter:
    x = np.linspace(-1000, 1000, 1000)
    z = r + x*(1j)
    Gamma = f_gamma(z)
    ax.plot(Gamma.real, Gamma.imag, c='blue', label=f"{r=}")

for x in x_iter:
    r = np.linspace(0, 1000, 1000)
    z = r + x*(1j)
    Gamma = f_gamma(z)
    ax.plot(Gamma.real, Gamma.imag, c='red', label=f"{x=}")

ax.set_title("uwave prelab - Smith Chart")
ax.grid()
```

## Answers to Smith chart questions:

1. Gamma traces out a circle of radius 0.5 which touches the origin.
2. The r=0 curve is the outermost circle of radius 1. The physical interpretation of the r=0 curve is zero resistance, so either termination with a closed circuit or a purely reactive component (inductor or capacitor). The reflection coefficient always has magnitude 1, and the reflected phase varies with the reactance x.
4. The constant-reactance curves are partial circles because the resistance does not go below 0. Negative resistance would imply charge spontaneously moving against a potential difference and gaining energy, which is unphysical for a passive component (though possible if the energy is sunk into a battery).
